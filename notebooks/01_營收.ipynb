# 營收資料：來自 MOPS 的月營收

取得並分析台灣全市場各板塊的月營收資料。

```python
import pandas as pd
from twmops import RevenueFetcher
import asyncio
```

## 市場代碼

- `sii`: 上市（TWSE 上市公司）
- `otc`: 上櫃（OTC）
- `rotc`: 興櫃（Emerging）
- `pub`: 公開發行（Public）

```python
def fetch_market_revenue():
    fetcher = RevenueFetcher()

    # 取得 TWSE (sii) 2024 年 3 月的營收資料
    revenues = fetcher.get_market_revenue(
        year=113,
        month=3,
        market="sii"
    )

    # 轉換為 DataFrame
    df = pd.DataFrame([
        {
            'stock_id': r.stock_id,
            'company': r.company_name,
            'revenue': r.revenue,
            'yoy_change': r.yoy_change,
            'accumulated': r.accumulated_revenue,
        }
        for r in revenues if r.revenue
    ])

    print(f"2024 年 3 月營收前 10 大:")
    print(df.nlargest(10, 'revenue')[['stock_id', 'company', 'revenue', 'yoy_change']])

    print(f"
年增率前 10 大:")
    print(df.nlargest(10, 'yoy_change')[['stock_id', 'company', 'yoy_change', 'revenue']])

    return df

df = fetch_market_revenue()
```

## 追蹤單一公司的營收變化

```python
def track_company_revenue():
    fetcher = RevenueFetcher()

    # 追蹤台積電 (2330) 2024 年 1-3 月營收
    months = []
    for month in [1, 2, 3]:
        rev = fetcher.get_single_revenue("2330", year=113, month=month)
        if rev:
            months.append({
                'month': month,
                'revenue': rev.revenue,
                'mom_change': rev.mom_change,
                'accumulated': rev.accumulated_revenue,
            })

    df = pd.DataFrame(months)
    print("台積電 (2330) 2024 年第 1 季營收:")
    print(df)

    return df

tsmc_trend = track_company_revenue()
```

## 欄位參考

每個 `MonthlyRevenue` 包含：
- `stock_id`: 股票代號（例如 "2330"）
- `company_name`: 公司官方名稱
- `year`, `month`: 民國年份與月份
- `revenue`: 本月營收（新台幣千元）
- `revenue_last_month`, `revenue_last_year`: 歷史值
- `mom_change`: 月增率（%）
- `yoy_change`: 年增率（%）
- `accumulated_revenue`: 累計營收（年度至今）
- `accumulated_last_year`: 去年同期累計營收
- `comment`: MOPS 備註
