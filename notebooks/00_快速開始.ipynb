````python
from twmops import RevenueFetcher, FinancialFetcher, DividendFetcher
````

## 取得月營收

取得所有上市公司（TWSE）2024 年 3 月的營收資料。

````python
def fetch_revenue_demo():
    fetcher = RevenueFetcher()

    # 取得 TWSE (sii) 2024 年 3 月的所有營收資料（民國年 113）
    revenues = fetcher.get_market_revenue(year=113, month=3, market="sii")

    # 顯示前 5 家公司
    print(f"找到 {len(revenues)} 家公司")
    for rev in revenues[:5]:
        print(f"  {rev.stock_id} {rev.company_name}: NT${rev.revenue:,} 千元")

    return revenues

# 執行
revenues = fetch_revenue_demo()
````

## 取得單一公司營收

````python
def fetch_tsmc_revenue():
    fetcher = RevenueFetcher()

    # 取得台積電 (2330) 月營收
    tsmc = fetcher.get_single_revenue("2330", year=113, month=3)

    if tsmc:
        print(f"{tsmc.company_name} ({tsmc.stock_id})")
        print(f"  2024 年 3 月營收: NT${tsmc.revenue:,} 千元")
        print(f"  年增率: {tsmc.yoy_change:.2f}%")
    return tsmc

tsmc_revenue = fetch_tsmc_revenue()
````

## 取得股利紀錄

取得公司的股利歷史。

````python
def fetch_tsmc_dividends():
    fetcher = DividendFetcher()

    # 取得台積電民國 110-113 年（2021-2024）的股利紀錄
    dividends = fetcher.get_dividends("2330", year_start=110, year_end=113)

    print(f"{dividends.company_name} - {len(dividends.records)} 筆股利紀錄")
    for record in dividends.records[:5]:
        print(f"  民國 {record.year} 第{record.quarter}季: "
              f"現金股利 NT${record.cash_dividend or 0:.2f}, "
              f"股票股利 NT${record.stock_dividend or 0:.2f}")

    return dividends

tsmc_dividends = fetch_tsmc_dividends()
````

## 取得財務報表

下載並解析 XBRL 資產負債表、損益表或現金流量表。

````python
def fetch_tsmc_balance_sheet():
    fetcher = FinancialFetcher()

    # 取得台積電 2024 年第 3 季資產負債表（民國 113 年第 3 季）
    stmt = fetcher.get_financial_statement(
        stock_id="2330",
        year=113,
        quarter=3,
        report_type="balance_sheet",
        format="tree"  # 分層結構
    )

    print(f"{stmt.stock_id} {stmt.report_type} - 第{stmt.quarter}季 民國{stmt.year}")
    print(f"  頂層項目: {len(stmt.items)}")
    for item in stmt.items[:3]:
        print(f"    {item.account_name}: {item.value}")

    return stmt

balance_sheet = fetch_tsmc_balance_sheet()
````
