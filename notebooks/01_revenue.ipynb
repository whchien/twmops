# Revenue Data: Monthly Revenue from MOPS

Fetch and analyze Taiwan monthly revenue data across all markets and sectors.

```python
import pandas as pd
from twmops import RevenueFetcher
import asyncio
```

## Markets

- `sii`: 上市 (TWSE listed)
- `otc`: 上櫃 (OTC)
- `rotc`: 興櫃 (Emerging)
- `pub`: 公開發行 (Public)

```python
def fetch_market_revenue():
    fetcher = RevenueFetcher()
    
    # Fetch TWSE listed companies for March 2024
    revenues = fetcher.get_market_revenue(
        year=113,
        month=3,
        market="sii"
    )
    
    # Convert to DataFrame
    df = pd.DataFrame([
        {
            'stock_id': r.stock_id,
            'company': r.company_name,
            'revenue': r.revenue,
            'yoy_change': r.yoy_change,
            'accumulated': r.accumulated_revenue,
        }
        for r in revenues if r.revenue
    ])
    
    print(f"Top 10 by March 2024 revenue:")
    print(df.nlargest(10, 'revenue')[['stock_id', 'company', 'revenue', 'yoy_change']])
    
    print(f"\nTop 10 by YoY growth:")
    print(df.nlargest(10, 'yoy_change')[['stock_id', 'company', 'yoy_change', 'revenue']])
    
    return df

df = fetch_market_revenue()
```

## Track a Single Company Over Time

```python
def track_company_revenue():
    fetcher = RevenueFetcher()
    
    # Track TSMC (2330) revenue from Jan–Mar 2024
    months = []
    for month in [1, 2, 3]:
        rev = fetcher.get_single_revenue("2330", year=113, month=month)
        if rev:
            months.append({
                'month': month,
                'revenue': rev.revenue,
                'mom_change': rev.mom_change,
                'accumulated': rev.accumulated_revenue,
            })
    
    df = pd.DataFrame(months)
    print("TSMC (2330) Revenue Q1 2024:")
    print(df)
    
    return df

tsmc_trend = track_company_revenue()
```

## Field Reference

Each `MonthlyRevenue` has:
- `stock_id`: Stock symbol (e.g., "2330")
- `company_name`: Official company name
- `year`, `month`: ROC year and month
- `revenue`: Current month revenue (thousands TWD)
- `revenue_last_month`, `revenue_last_year`: Historical values
- `mom_change`: Month-over-month change (%)
- `yoy_change`: Year-over-year change (%)
- `accumulated_revenue`: YTD cumulative revenue
- `accumulated_last_year`: YTD last year for comparison
- `comment`: Any notes from MOPS
