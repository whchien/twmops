# Insider Share Pledging: Director and Supervisor Holdings

Track director and supervisor share pledging activities from MOPS.

```python
from twmops import InsidersFetcher
import pandas as pd
import asyncio
```

## Fetch Pledging Details for a Company

Get detailed pledging records for individual directors and supervisors.

```python
def fetch_pledging_details():
    fetcher = InsidersFetcher()
    
    # Fetch TSMC (2330) share pledging records (using March 2026)
    result = fetcher.get_share_pledging("2330", year=115, month=3)
    
    print(f"{result.company_name} ({result.stock_id})")
    print(f"Report date: {result.year}/{result.month}")
    print(f"Total pledging records: {len(result.details)}")
    print()
    
    # Show pledging details
    for detail in result.details[:10]:
        print(f"Person: {detail.name}")
        print(f"  Title: {detail.title}")
        if detail.pledged_shares is not None:
            print(f"  Pledged shares: {detail.pledged_shares:,}")
        if detail.current_shares is not None:
            print(f"  Current shares: {detail.current_shares:,}")
        if detail.pledge_ratio is not None:
            print(f"  Pledge ratio: {detail.pledge_ratio:.2%}")
        print()
    
    if len(result.details) > 10:
        print(f"... and {len(result.details) - 10} more records")
    
    return result

pledging_response = fetch_pledging_details()
```

## Analyze Pledging Summary

Get aggregated pledging statistics by title/position.

```python
def analyze_pledging_summary():
    fetcher = InsidersFetcher()
    
    result = fetcher.get_share_pledging("2330", year=115, month=3)
    
    print(f"Pledging Summary for {result.company_name}")
    print(f"Report date: {result.year}/{result.month}")
    print()
    
    # Group by title
    df = pd.DataFrame([
        {
            'name': detail.name,
            'title': detail.title,
            'current_shares': detail.current_shares,
            'pledged_shares': detail.pledged_shares,
            'pledge_ratio': detail.pledge_ratio,
        }
        for detail in result.details
    ])
    
    # Summary by title
    print("Summary by Title:")
    summary = df.groupby('title').agg({
        'current_shares': 'sum',
        'pledged_shares': 'sum',
        'pledge_ratio': 'mean',
    }).round(4)
    summary['count'] = df.groupby('title').size()
    print(summary)
    print()
    
    # Top pledgers (filter out None values)
    print("Top 5 Pledgers (by pledge_ratio):")
    df_with_ratio = df[df['pledge_ratio'].notna()]
    if len(df_with_ratio) > 0:
        top_pledgers = df_with_ratio.nlargest(5, 'pledge_ratio')[['name', 'title', 'pledged_shares', 'current_shares', 'pledge_ratio']]
        print(top_pledgers.to_string(index=False))
    else:
        print("No pledging ratio data available")
    
    return df

pledging_df = analyze_pledging_summary()
```

## Field Reference

Each pledging record has:
- `name`: Director/supervisor name
- `title`: Position (董事長, 董事, 監察人, etc.)
- `current_shares`: Number of shares currently owned
- `pledged_shares`: Number of shares pledged as collateral
- `pledge_ratio`: Pledged shares / Current shares
